```python
import pandas as pd
import numpy as np
import ast
from tqdm import tqdm
from sentence_transformers import SentenceTransformer
import os
import matplotlib.pyplot as plt


embedder = SentenceTransformer(
    "mixedbread-ai/mxbai-embed-large-v1"
)


def json_to_text(json_str):

    try:
        d = ast.literal_eval(json_str)
        if isinstance(d, dict):
            return " | ".join(f"{k}: {v}" for k, v in d.items())
        return str(d)
    except Exception:
        return str(json_str)


def batch_embed(texts, batch_size=32, desc="Embeddings"):

    embs = []
    for i in tqdm(range(0, len(texts), batch_size), desc=desc):
        batch = texts[i:i+batch_size]
        e = embedder.encode(
            batch,
            batch_size=len(batch),
            convert_to_tensor=True,
            normalize_embeddings=True
        )
        embs.append(e.cpu().numpy())
    return np.vstack(embs)


def _plot_and_save_hist(sims, title, out_path):

    plt.figure()
    plt.hist(sims, bins=40)
    plt.title(title)
    plt.xlabel("Similaridade de cosseno")
    plt.ylabel("Frequência")
    plt.tight_layout()
    plt.savefig(out_path)
    plt.close()


def evaluate_json_bench_vs_text(df_bench, dataset_name, out_dir="eval_bench_vs_text"):
    os.makedirs(out_dir, exist_ok=True)

    subset = df_bench[df_bench["dataset"] == dataset_name].copy()

    subset["json_bench_text"] = subset["json_answer"].apply(json_to_text)
    subset["text"] = subset["text"].astype(str)

    json_texts = subset["json_bench_text"].tolist()
    text_texts = subset["text"].tolist()

    emb_json = batch_embed(
        json_texts, 32, desc=f"Embeddings (JSON benchmark - {dataset_name})")
    emb_text = batch_embed(
        text_texts, 32, desc=f"Embeddings (TEXT benchmark - {dataset_name})")

    sims = np.sum(emb_json * emb_text, axis=1)

    np.save(os.path.join(out_dir, f"{dataset_name}_emb_bench.npy"), emb_json)
    np.save(os.path.join(out_dir, f"{dataset_name}_emb_text.npy"), emb_text)
    np.save(os.path.join(out_dir, f"{dataset_name}_similarity.npy"), sims)
    subset.to_parquet(os.path.join(
        out_dir, f"{dataset_name}_bench_pairs.parquet"), index=False)

    _plot_and_save_hist(
        sims,
        title=f"{dataset_name} – JSON benchmark vs texto",
        out_path=os.path.join(out_dir, f"{dataset_name}_similarity_hist.png")
    )

    # métricas
    print(f"\n[Benchmark] Dataset: {dataset_name}")
    print(f"Média:   {np.mean(sims):.4f}")
    print(f"Mediana: {np.median(sims):.4f}")
    print(f"Std:     {np.std(sims):.4f}")
    print(f"Min:     {np.min(sims):.4f}")
    print(f"Max:     {np.max(sims):.4f}")

    return sims, subset, emb_json, emb_text


def evaluate_json_pred_vs_text(df_lx, dataset_name, out_dir="eval_pred_vs_text"):
    os.makedirs(out_dir, exist_ok=True)

    print(f"\n[Pred] Nosso JSON vs Texto – Dataset: {dataset_name}")

    subset = df_lx[df_lx["dataset"] == dataset_name].copy()

    subset["text"] = subset["text"].astype(str)
    subset["json_pred_text"] = subset["json_answer"].apply(json_to_text)

    pred_texts = subset["json_pred_text"].tolist()
    text_texts = subset["text"].tolist()

    emb_pred = batch_embed(
        pred_texts, 32, desc=f"Embeddings (JSON nosso - {dataset_name})")
    emb_text = batch_embed(
        text_texts, 32, desc=f"Embeddings (TEXT nosso - {dataset_name})")

    sims = np.sum(emb_pred * emb_text, axis=1)

    np.save(os.path.join(out_dir, f"{dataset_name}_emb_pred.npy"), emb_pred)
    np.save(os.path.join(out_dir, f"{dataset_name}_emb_text.npy"), emb_text)
    np.save(os.path.join(out_dir, f"{dataset_name}_similarity.npy"), sims)
    subset.to_parquet(os.path.join(
        out_dir, f"{dataset_name}_pred_pairs.parquet"), index=False)

    _plot_and_save_hist(
        sims,
        title=f"{dataset_name} – JSON nosso vs texto",
        out_path=os.path.join(out_dir, f"{dataset_name}_similarity_hist.png")
    )

    print(f"[Pred] Dataset: {dataset_name}")
    print(f"Média:   {np.mean(sims):.4f}")
    print(f"Mediana: {np.median(sims):.4f}")
    print(f"Std:     {np.std(sims):.4f}")
    print(f"Min:     {np.min(sims):.4f}")
    print(f"Max:     {np.max(sims):.4f}")

    return sims, subset, emb_pred, emb_text


def run_full_evaluation(df_bench, df_lx):

    datasets = ["mave", "ae-110k", "oa-mine"]
    results = {}

    for ds in datasets:
        print(f"PROCESSANDO DATASET: {ds}")

        # A: benchmark json_answer vs texto (df_bench)
        sims_b, pairs_b, emb_json_b, emb_text_b = evaluate_json_bench_vs_text(
            df_bench, ds)

        # B: nosso json_answer vs texto (df_lx)
        sims_p, pairs_p, emb_json_p, emb_text_p = evaluate_json_pred_vs_text(
            df_lx, ds)

        results[ds] = {
            "bench": {
                "sims": sims_b,
                "pairs": pairs_b,
                "emb_json": emb_json_b,
                "emb_text": emb_text_b
            },
            "pred": {
                "sims": sims_p,
                "pairs": pairs_p,
                "emb_json": emb_json_p,
                "emb_text": emb_text_p
            }
        }

        if sims_b is not None and sims_p is not None:
            print(f"    Média benchmark: {np.mean(sims_b):.4f}")
            print(f"    Média nosso:     {np.mean(sims_p):.4f}")

    return results
```
