```python
""" Functional JSON """

from transformers import AutoModelForCausalLM, AutoTokenizer

# Load model and tokenizer
model_id = "LiquidAI/LFM2-350M-Extract"
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto",
    dtype="bfloat16",
    #    attn_implementation="flash_attention_2" <- uncomment on compatible GPU
)
tokenizer = AutoTokenizer.from_pretrained(model_id)

# Create message
system_prompt = """Identify and extract the following information. Present as a JSON object.

Proposal Details:
proposal_id: Unique identifier for the proposal.
client_name: Name of the client submitting the proposal.
project_title: Title or brief description of the logistics project.
total_project_cost: Total estimated cost of the project.

Transport Routes:
transport_routes: An array of route objects.
route_id: Unique identifier for each route.
origin: Starting location of the route.
destination: Ending location of the route.
mode: Transportation method (e.g., Air Freight, Ground Transport).
estimated_transit_time: Expected duration of the route.
carrier: Company responsible for transport.
cost: Cost associated with the route.

Equipment List:
equipment_list: An array of equipment objects.
item_id: Unique identifier for each item.
item_name: Description of the item.
quantity: Number of units of the item.
unit_cost: Cost per unit of the item.
weight: Weight of the item.
special_handling: Any specific requirements for handling the item.

Timeline:
timeline: An object containing project dates.
project_start_date: Date the project is scheduled to begin.
phase_1_completion_date: Completion date for the first project phase.
phase_2_completion_date: Completion date for the second project phase.
project_end_date: Overall project completion date.

Risk & Insurance:
risk_assessment_narrative: Description of potential risks associated with the project.
contingency_planning_notes: Notes on plans to mitigate identified risks.
insurance_details: An object containing insurance information.
policy_number: Insurance policy number.
insurance_provider: Name of the insurance provider.
coverage_amount: Amount of insurance coverage.
coverage_type: Type of insurance coverage."""

user_prompt = """Source Document:
---
Logistics Solutions, Inc.
14 Nightingale Lane, Oxford, OX1 4AX, UK
Date: October 27, 2024

Proposal: Quantum Entanglement Device – Secure Transit
Proposal ID: LOG-2024-10-27-001

To: Aethelred Technologies

Introduction
Logistics Solutions, Inc. is pleased to submit this proposal for the secure and timely transit of your Quantum Entanglement Device. We understand the critical nature of this shipment and the need for meticulous planning and execution. This document details our proposed approach, including transport routes, timelines, equipment handling procedures, risk assessments, and insurance coverage.

Project Overview
Project Title: Quantum Entanglement Device - Secure Transit
Total Project Cost: $1,250,000.00

Transport Routes
Our proposed transport plan involves a two-stage approach, combining air freight and specialized ground transport to ensure the utmost security and minimize potential disruptions.

Route 1: Geneva to Los Alamos
Route ID: RT-001
Origin: Geneva, Switzerland
Destination: Los Alamos, New Mexico, USA
Mode: Air Freight
Carrier: SwissAir
Estimated Transit Time: 72 hours
Cost: $350,000.00

Route 2: Los Alamos to Cambridge
Route ID: RT-002
Origin: Los Alamos, New Mexico, USA
Destination: Cambridge, UK
Mode: Specialized Ground Transport
Carrier: SecureLogix
Estimated Transit Time: 120 hours
Cost: $500,000.00

Equipment List
The following equipment requires secure transport. Detailed handling instructions are provided below.

Item ID: EQ-001
Item Name: Quantum Core Unit
Quantity: 1
Unit Cost: $800,000.00
Weight: 15 kg
Special Handling: Temperature-controlled, vibration isolated

Item ID: EQ-002
Item Name: Cryogenic Cooling System
Quantity: 1
Unit Cost: $100,000.00
Weight: 50 kg
Special Handling: Nitrogen refill capabilities required

Project Timeline
Project Start Date: 2024-11-15
Phase 1 Completion Date: 2024-11-22 (Air Freight to Los Alamos)
Phase 2 Completion Date: 2024-11-29 (Preparation for Ground Transport)
Project End Date: 2024-12-06 (Delivery to Cambridge)

Risk Assessment & Contingency Planning
Risk Assessment Narrative: The fragility of the Quantum Core demands meticulous handling. Potential disruptions, like atmospheric turbulence or unforeseen ground transport delays, could compromise the device. We must consider the 'shadows' that may fall across our path.

Contingency Planning Notes: Should the primary air route be unavailable, a secondary route via Iceland is prepared. For ground transport, redundant GPS tracking and a dedicated security escort will be utilized. The 'still, small voice' of caution guides our preparations.

Insurance Details
Policy Number: INS-456789
Insurance Provider: GlobalSecure Insurance
Coverage Type: All-Risk
Coverage Amount: $2,000,000.00

We are confident that Logistics Solutions, Inc. possesses the expertise and resources to ensure the safe and secure transit of your valuable equipment. We look forward to the opportunity to serve you.

Sincerely,

Alastair Finch
Director of Operations
Logistics Solutions, Inc."""
message = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]

# Generate answer
input_ids = tokenizer.apply_chat_template(
    message,
    add_generation_prompt=True,
    return_tensors="pt",
    tokenize=True,
).to(model.device)

output = model.generate(
    input_ids,
    do_sample=False,
    max_new_tokens=1024,
)

print(tokenizer.decode(output[0], skip_special_tokens=False))
```
