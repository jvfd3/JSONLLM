```python
from Modules.evaluation import get_distribution_metrics
from Modules.plotting_charts import plot_results
import pandas as pd
# import matplotlib.pyplot as plt
# import openpyxl
# import seaborn as sns
```

```python

jsonllm_eval_df_path = 'Datasets/jsonllm/jsonllm_evaluation_metrics.parquet'

# Load df
jsonllm_eval_df = pd.read_parquet(jsonllm_eval_df_path)
jsonllm_eval_df.head()
```

```python
# jsonllm_eval_df

metrics = {}

general_metrics = {}

general_metrics['text_vs_gt'] = get_distribution_metrics(
    jsonllm_eval_df['text_vs_gt_similarity_score'])
general_metrics['text_vs_lx'] = get_distribution_metrics(
    jsonllm_eval_df['text_vs_lx_similarity_score'])
general_metrics['gt_vs_lx'] = get_distribution_metrics(
    jsonllm_eval_df['gt_vs_lx_similarity_score'])

metrics['general'] = general_metrics

# MAVE

mave_df_eval = jsonllm_eval_df[jsonllm_eval_df['dataset'] == 'mave']
mave_metrics = {}
mave_metrics['text_vs_gt'] = get_distribution_metrics(
    mave_df_eval['text_vs_gt_similarity_score'])
mave_metrics['text_vs_lx'] = get_distribution_metrics(
    mave_df_eval['text_vs_lx_similarity_score'])
mave_metrics['gt_vs_lx'] = get_distribution_metrics(
    mave_df_eval['gt_vs_lx_similarity_score'])

metrics['mave'] = mave_metrics

# ae-110k

ae110k_df_eval = jsonllm_eval_df[jsonllm_eval_df['dataset'] == 'ae-110k']
ae110k_metrics = {}
ae110k_metrics['text_vs_gt'] = get_distribution_metrics(
    ae110k_df_eval['text_vs_gt_similarity_score'])
ae110k_metrics['text_vs_lx'] = get_distribution_metrics(
    ae110k_df_eval['text_vs_lx_similarity_score'])
ae110k_metrics['gt_vs_lx'] = get_distribution_metrics(
    ae110k_df_eval['gt_vs_lx_similarity_score'])

metrics['ae-110k'] = ae110k_metrics

# oa-mine

oa_mine_df_eval = jsonllm_eval_df[jsonllm_eval_df['dataset'] == 'oa-mine']
oa_mine_metrics = {}
oa_mine_metrics['text_vs_gt'] = get_distribution_metrics(
    oa_mine_df_eval['text_vs_gt_similarity_score'])
oa_mine_metrics['text_vs_lx'] = get_distribution_metrics(
    oa_mine_df_eval['text_vs_lx_similarity_score'])
oa_mine_metrics['gt_vs_lx'] = get_distribution_metrics(
    oa_mine_df_eval['gt_vs_lx_similarity_score'])

metrics['oa-mine'] = oa_mine_metrics

# ===

# Convert them into a dataframe for better visualization (Mean, Median, Std, Min, Max) x (General, MAVE, AE-110k, OA-Mine)
metrics_df = pd.DataFrame()
for dataset_name, dataset_metrics in metrics.items():
    for metric_name, metric_values in dataset_metrics.items():
        for stat_name, stat_value in metric_values.items():
            metrics_df.loc[stat_name,
                           f"{dataset_name}_{metric_name}"] = stat_value
metrics_df.to_excel('Datasets/jsonllm/jsonllm_evaluation_metrics_summary.xlsx')
metrics_df
```

```python
""" Plotting Charts """

plot_results(df=mave_df_eval, especific='mave')
plot_results(df=ae110k_df_eval, especific='ae110k')
plot_results(df=oa_mine_df_eval, especific='oa-mine')
```
